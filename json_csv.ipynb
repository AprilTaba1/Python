```python
import pandas as pd

import json 
with open('Desktop/mva_zoom_attendance_group_2.json') as f:
    d = json.load(f)
    print(d)
```

```python
if isinstance(d,list):
    df = pd.json_normalize(d)
elif  isinstance(d,dict):
    df = pd.json_normalize([d])
else:
    raise ValueError('not working') 
```

```python
pd.set_option('display.max_columns', None)
```

```python
print(df)
```

```python
df.to_csv('zoom_data.csv', index = False)
```
